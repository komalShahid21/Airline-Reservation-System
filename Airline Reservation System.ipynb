```python
def registration(email_password):
    print("REGISTRATION")

    #Taking email and password and checking email format using exception 
    #and this loop will continue until user don't enter valid email
    
    while 'True':
        email=input("Enter your Email ID: ")
        password=input("Enter password: ")
        
        try:
            if '@' not in email:
                raise ValueError("Entered Email is invalid")
            else:
                break
        except ValueError as VE:
            print(VE)

    email_password[email]=password

    file=open('registration.txt','w')
    file.write(str(email_password))
    file.close()

    return email_password
```

```python
def login():
    print("")
    print("LOGIN PAGE")
    
    #Taking email and password from user
    
    email=input("Enter your Email ID: ")
    password=input("Enter password: ")

    file=open('registration.txt','r')
    cont=file.read()
    file.close()
    contant=eval(cont)

    flag=0
     
    #This loop will continue until it match user entered email 
    #and password with all which are present in registartion file
    
    for key,value in contant.items():
        if key==email and value==password:
            flag=0
            print("LOGIN SUCCESSFULLY")
            break
        else:
            flag=1
    
    if flag==1:
        print("Entered Email or password is incorrect")
```

```python
def profile_management():
    print("")
    print("PERSONAL ACCOUNT DETAILS")
    gender=input("Gender: ")
    title=input("Title: ")
    name=input("Full Name: ")

    # This loop will continue until the date of birth entered by user is not of format defined
    # and date, month and year are not correct. On wrong format or date it shall display the error message
    
    while 'True':
        date_of_birth=input("Date of Birth(dd/mm/yyyy): ")
        
        if date_of_birth[2]!='/' or date_of_birth[5]!='/':
            print("Invalid date format")
        elif int(date_of_birth[0:2])>31 or int(date_of_birth[3:5])>12 or (1909>int(date_of_birth[6:10]) or int(date_of_birth[6:10])>2009):
            print("Entered date is invalid")
        else:
            break

    nationality=input("Nationality: ")

    # This loop will continue until the passport number is not correct and display the error with message to correct it
    
    while 'True':
        passport_num= input("Passport Number: ")
        
        if len(passport_num) > 9 or (not(passport_num[0:2]).isalpha()) or (not(passport_num[2:9]).isnumeric()):
            print("Entered passport number is invalid")
            print("Passport number should start with 2 alphabets followed by 7 numbers")
        else:
            break

    # This loop will continue until the phone number is not correct and display the error with message to correct it
    
    while 'True':
        phone_num=input("Phone Number: ")
        
        if len(phone_num)!=11:
            print("Entered number is not correct")
        else:
            break

    address=input("Preferred Address: ")
    country=input("Country of Residence: ")
    city=input("City: ")
    state=input("State: ")

    #This loop will continue until the postal code is not correct 
    # and display the error message user entered more than 5 digit in it
    
    while 'True':
        postal_code=input("Postal Code: ")
        
        if len(postal_code) > 5:
            print("Entered postal code is not correct")
        else:
            break

    profile_management_dict={'gender':gender, 'title':title, 'name':name, 'date_of_birth':date_of_birth, 'nationality': nationality, 'passport_num':passport_num, 'phone_num':phone_num, 'address':address, 'country':country, 'city':city, 'state':state}

    file=open('profile_management.txt','a')
    file.write(str(profile_management_dict) + '\n')
    file.close()
```

```python
#drivers code
#Profile management that display all error message on wrong info entered by user

profile_management()
```

```python
#drivers code
#Profile management that display correct information entered by user

profile_management()
```
